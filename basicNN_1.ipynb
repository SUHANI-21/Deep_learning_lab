```python
import numpy as np

# 1. Define Activation Functions
def threshold(x):
    return np.where(x >= 0, 1, 0)

def relu(x):
    return np.maximum(0, x)

def sigmoid(x):
    return 1 / (1 + np.exp(-x))

def tanh(x):
    return np.tanh(x)

# 2. Setup Inputs and Weights (Simple 2x2 setup)
inputs = np.array([0.5, -0.2])
weights = np.array([0.8, 0.4])

# Model 1: Fixed Bias (e.g., 0.1)
# Using Threshold and Sigmoid
fixed_bias = 0.1
z_fixed = np.dot(inputs, weights) + fixed_bias
print(f"Fixed Bias Model (Output):")
print(f"- Threshold: {threshold(z_fixed)}")
print(f"- Sigmoid:   {sigmoid(z_fixed):.4f}\n")

# Model 2: Varying Bias (e.g., -0.5)
# Using ReLU and Tanh
varying_bias = -0.5
z_varying = np.dot(inputs, weights) + varying_bias
print(f"Varying Bias Model (Output):")
print(f"- ReLU: {relu(z_varying)}")
print(f"- Tanh: {tanh(z_varying):.4f}")
```
