```python
import numpy as np

# XOR dataset
X = np.array([[0,0],
              [0,1],
              [1,0],
              [1,1]])

y = np.array([[0],[1],[1],[0]])

# Activation functions
def sigmoid(x):
    return 1 / (1 + np.exp(-x))

def sigmoid_derivative(x):
    return x * (1 - x)

# Network architecture
input_size = 2
hidden_size = 2
output_size = 1

# Initialize weights
np.random.seed(42)
W1 = np.random.randn(input_size, hidden_size) * 0.5
b1 = np.zeros((1, hidden_size))
W2 = np.random.randn(hidden_size, output_size) * 0.5
b2 = np.zeros((1, output_size))

learning_rate = 0.5
epochs = 10000

for epoch in range(epochs):
    # Forward pass
    z1 = X @ W1 + b1
    a1 = sigmoid(z1)
    z2 = a1 @ W2 + b2
    y_hat = sigmoid(z2)

    # Binary Cross-Entropy loss
    y_hat_clipped = np.clip(y_hat, 1e-7, 1-1e-7)  # avoid log(0)
    loss = -np.mean(y * np.log(y_hat_clipped) + (1 - y) * np.log(1 - y_hat_clipped))

    # Backpropagation
    # For BCE + sigmoid, derivative simplifies
    d_output = y_hat - y

    dW2 = a1.T @ d_output
    db2 = np.sum(d_output, axis=0, keepdims=True)

    d_hidden = d_output @ W2.T * sigmoid_derivative(a1)

    dW1 = X.T @ d_hidden
    db1 = np.sum(d_hidden, axis=0, keepdims=True)

    # Update weights
    W2 -= learning_rate * dW2
    b2 -= learning_rate * db2
    W1 -= learning_rate * dW1
    b1 -= learning_rate * db1

    if epoch % 1000 == 0:
        print(f"Epoch {epoch}, Loss: {loss:.6f}")

# Final predictions
print("\nPredictions after training:")
print(np.round(y_hat))
```
